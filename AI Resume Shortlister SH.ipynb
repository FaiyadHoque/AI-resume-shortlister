```python
##dropped columns that have null value greater than 60%

null_percentages = df.isnull().sum() / df.shape[0] * 100
columns_to_drop = null_percentages[null_percentages > 60].index
df_dropped = df.drop(columns=columns_to_drop)
display(df_dropped.head())
```

```python
for column in df_dropped.columns:
    if df_dropped[column].dtype == 'object':
        df_dropped[column].fillna('Unknown', inplace=True)

# You can also use other methods like forward fill (ffill), backward fill (bfill), or fill with mean/median/mode for numerical columns.
# For example:
# df_dropped['numerical_column'].fillna(df_dropped['numerical_column'].mean(), inplace=True)

display(df_dropped.isnull().sum())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Combine the text from the specified columns into a single column
df_dropped['combined_text'] = df_dropped['career_objective'].fillna('') + ' ' + \
                             df_dropped['skills'].fillna('') + ' ' + \
                             df_dropped['responsibilities'].fillna('')

# Initialize TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer(stop_words='english', max_features=5000) # You can adjust max_features

# Fit and transform the combined text
tfidf_matrix = tfidf_vectorizer.fit_transform(df_dropped['combined_text'])

# You can now use tfidf_matrix for further analysis (e.g., clustering, similarity calculations)
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
df_dropped.to_csv('cleaned_resume_data.csv', index=False)
print("Cleaned dataset saved to 'cleaned_resume_data.csv'")
```

```python
cleaned_df = pd.read_csv('cleaned_resume_data.csv')
display(cleaned_df.head())
display(cleaned_df.info())
display(cleaned_df.isnull().sum())
```

```python
# Display the head of the dataframe
display(cleaned_df.head())

# Display descriptive statistics for numerical columns
display(cleaned_df.describe())

# Display descriptive statistics for object columns (including counts of unique values)
display(cleaned_df.describe(include='object'))
```

```python
# Analyze value counts for some categorical columns
display(cleaned_df['﻿job_position_name'].value_counts().head(10))
display(cleaned_df['educationaL_requirements'].value_counts().head(10))
display(cleaned_df['experiencere_requirement'].value_counts().head(10))
```

```python
y = cleaned_df['matched_score']
X = cleaned_df.drop('matched_score', axis=1)
display(X.head())
display(y.head())
```

```python
# Identify categorical and numerical columns
categorical_cols = X.select_dtypes(include='object').columns.tolist()
# Exclude 'combined_text' from categorical encoding as we'll use its TF-IDF representation
categorical_cols.remove('combined_text')

# Apply one-hot encoding to categorical features
X_categorical = pd.get_dummies(X[categorical_cols], dummy_na=False)

# Ensure indices align before concatenating
X_categorical.index = cleaned_df.index

# Concatenate one-hot encoded features with TF-IDF matrix
X_processed = pd.concat([X_categorical, pd.DataFrame(tfidf_matrix.toarray(), index=cleaned_df.index)], axis=1)

display(X_processed.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_processed, y, test_size=0.2, random_state=42)

print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
from xgboost import XGBRegressor

```

```python
# Convert all column names to strings and then clean
X_train.columns = [str(col).replace('[', '').replace(']', '').replace('<', '') for col in X_train.columns]
X_test.columns = [str(col).replace('[', '').replace(']', '').replace('<', '') for col in X_test.columns]

# Initialize the XGBoost Regressor model
xgb_model = XGBRegressor(random_state=42)

# Train the model
xgb_model.fit(X_train, y_train)
```

```python
y_train_pred = xgb_model.predict(X_train)
y_test_pred = xgb_model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# Calculate evaluation metrics for training set
mse_train = mean_squared_error(y_train, y_train_pred)
rmse_train = np.sqrt(mse_train)
mae_train = mean_absolute_error(y_train, y_train_pred)
r2_train = r2_score(y_train, y_train_pred)

# Calculate evaluation metrics for testing set
mse_test = mean_squared_error(y_test, y_test_pred)
rmse_test = np.sqrt(mse_test)
mae_test = mean_absolute_error(y_test, y_test_pred)
r2_test = r2_score(y_test, y_test_pred)

print("Training Set Metrics:")
print(f"  MSE: {mse_train:.4f}")
print(f"  RMSE: {rmse_train:.4f}")
print(f"  MAE: {mae_train:.4f}")
print(f"  R-squared: {r2_train:.4f}")

print("\nTesting Set Metrics:")
print(f"  MSE: {mse_test:.4f}")
print(f"  RMSE: {rmse_test:.4f}")
print(f"  MAE: {mae_test:.4f}")
print(f"  R-squared: {r2_test:.4f}")
```
